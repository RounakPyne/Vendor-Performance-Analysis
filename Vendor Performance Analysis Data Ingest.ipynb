```python
# Ingest data in DB
def ingest_db(df, table_name, engine):
    df.to_sql(table_name, con = engine, if_exists = 'replace', index = False)
```

```python
# Ingest sales table in chunks
import pandas as pd

chunk_size = 50000
chunks = pd.read_csv("Data/sales.csv", chunksize=chunk_size)

for i, chunk in enumerate(chunks):
    chunk.to_sql("sales", engine, if_exists="append", index=False)
    print(f"Chunk {i+1} ingested — {(i+1)*chunk_size} rows done")

print("Sales table ingestion complete!")
```
